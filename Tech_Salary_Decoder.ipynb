```python
# Placing in Emote, hybrid and Onsite

work_mode_mapping = {
    "Remote": "Remote",
    "WFH": "Remote",
    "Hybrid": "Hybrid",
    "Hybrid (3 days)": "Hybrid",
    "Onsite": "Onsite",
    "Work from Office": "Onsite"
}
df["work_mode_clean"] = df["work_mode"].replace(work_mode_mapping)
print(df["work_mode_clean"].value_counts())
```

```python
#AI assisted: for checking the handling process of ctc in single parsing
# Converting different salary formats into numeric LPA values.

def parse_ctc(ctc):
    if pd.isna(ctc):
        return np.nan
    ctc = str(ctc).strip()
    ctc = ctc.replace("₹", "")
    ctc = ctc.replace("LPA", "")
    ctc = ctc.replace("lpa", "")
    ctc = ctc.strip()
    if "," in ctc:        # Handle Indian currency format like 15,50,000
        number = ctc.replace(",", "")
        if number.isdigit():
            number = float(number)
            if number > 1000:
                return number / 100000

        return pd.to_numeric(number, errors="coerce")
    return pd.to_numeric(ctc, errors="coerce")
```

```python
df["current_ctc_lpa"] = df["current_ctc"].apply(parse_ctc)
df["previous_ctc_lpa"] = df["previous_ctc"].apply(parse_ctc)
```

**Task-3**

**3.1: Calculating the median, mean, minimum, and maximum Current CTC (LPA) for each role**

```python
role_salary = (
    df.groupby("role_clean")["current_ctc_lpa"]
    .agg(
        Median_CTC="median",
        Mean_CTC="mean",
        Minimum_CTC="min",
        Maximum_CTC="max"
    )
    .sort_values(by="Median_CTC", ascending=False)
)
print(role_salary)
```

```python
# Displaying highest and lowest paying roles
highest_role = role_salary.index[0]
lowest_role = role_salary.index[-1]
print("Highest Paying Role:", highest_role)
print("Median CTC:", role_salary.iloc[0]["Median_CTC"], "LPA")
print()
print("Lowest Paying Role:", lowest_role)
print("Median CTC:", role_salary.iloc[-1]["Median_CTC"], "LPA")
```

```python
# Comparing mean and median salaries
role_salary["Difference"] = (role_salary["Mean_CTC"] - role_salary["Median_CTC"])
print(role_salary)
```

**3.2: Analysing how median salary changes**

```python
backend_df = df[df["role_clean"] == "SDE Backend"].copy()
backend_df["exp_band"] = pd.cut(
    backend_df["years_exp"],
    bins=[-1, 1, 3, 5, 100],
    labels=["0-1", "2-3", "4-5", "6+"]
)
backend_salary = (
    backend_df
    .groupby("exp_band")["current_ctc_lpa"]
    .median()
)
print(backend_salary)
```

```python
growth = backend_salary.pct_change() * 100
experience_table = pd.DataFrame({
    "Median_CTC": backend_salary,
    "Growth_%": growth
})
print(experience_table)
```

**3.3: Comparing median salary to identify which skill provide highest salary**

```python
sde_df = df[df["role_clean"].isin([
    "SDE Backend",
    "SDE Frontend",
    "SDE Full-Stack"
])].copy()
sde_df["skills"] = sde_df["skills"].fillna("")
skills = [
    "AWS",
    "ML",
    "System Design",
    "Kubernetes"
]
skill_results = []
for skill in skills:
    with_skill = sde_df[
        sde_df["skills"].str.contains(skill, case=False)
    ]["current_ctc_lpa"].median()
    without_skill = sde_df[
        ~sde_df["skills"].str.contains(skill, case=False)
    ]["current_ctc_lpa"].median()
    premium = ((with_skill - without_skill) / without_skill) * 100
    skill_results.append([
        skill,
        round(with_skill, 2),
        round(without_skill, 2),
        round(premium, 2)
    ])
skill_df = pd.DataFrame(
    skill_results,
    columns=[
        "Skill",
        "With Skill",
        "Without Skill",
        "Premium (%)"
    ]
)
print(skill_df)
```

**3.4: Calculating salary premium**

```python
backend_company = df[df["role_clean"] == "SDE Backend"]
company_salary = (
    backend_company
    .groupby("company_type_clean")["current_ctc_lpa"]
    .median()
    .reindex(["Unicorn", "MNC", "Mid-size", "Early-stage"])
)

print(company_salary)
```

```python

unicorn_salary = company_salary["Unicorn"]
mnc_salary = company_salary["MNC"]
premium = ((unicorn_salary - mnc_salary) / mnc_salary) * 100
print(f"Unicorn Median CTC : {unicorn_salary:.2f} LPA")
print(f"MNC Median CTC      : {mnc_salary:.2f} LPA")
print(f"Premium             : {premium:.2f}%")
```

**3.5: identifying professionals earning more below their peers**

```python
df["exp_band"] = pd.cut(
    df["years_exp"],
    bins=[-1, 1, 3, 5, 100],
    labels=["0-1", "2-3", "4-5", "6+"]
)
```

```python
group_size = (
    df.groupby(
        ["role_clean", "company_type_clean", "exp_band"]
    )["current_ctc_lpa"]
    .transform("count")
)
filtered_df = df[group_size >= 10].copy()
```

```python
filtered_df["group_median"] = (    #Calculating median
    filtered_df.groupby(
        ["role_clean", "company_type_clean", "exp_band"]
    )["current_ctc_lpa"]
    .transform("median")
)
```

```python
filtered_df["gap"] = (
    filtered_df["current_ctc_lpa"] -
    filtered_df["group_median"]
)
```

```python
# Displaying the ten most underpaid professionals.
underpaid = (
    filtered_df.sort_values("gap")
    [
        [
            "employee_id",
            "role_clean",
            "company_type_clean",
            "years_exp",
            "current_ctc_lpa",
            "group_median",
            "gap"
        ]
    ].head(10)
)
print(underpaid)
```
